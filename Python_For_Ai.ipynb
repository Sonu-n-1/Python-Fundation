# Match / Case

```python

status = "PENDING"
match status:
    case "SUCCESS":
        print("Payment done")
    case "PENDING":
        print("Waiting for your bank...")
    case "FAILED":
        print("Payment failed. Money will be refunded.")
    case _:
        print("Unknown status")
              
```

# Generator Functions

#### It uses yield to produce values one at a time rather than returning them all at once

```python
def count(n):
    i = 1
    while i <= n:
        yield i
        i+=1
for num in count(5):
    print(num)
```

```python
# AN EMI CALCULATOR USING FUNCTIONS

def emi (principal,annual_rate,months):
    r = annual_rate / 12 / 100
    factor = (1+r)**months
    return round(principal * r * factor / (factor - 1))
print("Laptop:",emi(60000,12,12))
print("bike:",emi(90000,10,24))
```

# 4 Pillars of OOPs
- Encapsulation: Protect and manage data.

- Inheritance: Reuse and extend code.

- Polymorphism: Same interface, different behavior.

- Abstraction: Hide implementation details.

- Encapsulation - Protect data inside the object  (self.__balance)
- Inheritance - Reuse code from a parent class  (class Teacher(Person))
- Polymorphism - One action, many forms  (method.pay(500))
- Abstraction - Hide details, show what matters  (@abstractmethod)

```python
# Encapsulation

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())
```

```python
# Inheritance

class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

dog = Dog()
dog.eat()
dog.bark()
```

```python
# Method overriding - A child class can also provide its own implementation of a parent method.

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

dog = Dog()
dog.sound()
# commonly used in implimenting polymorphism
```

```python
# Polymorphism

class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

def make_sound(animal):
    animal.sound()

make_sound(Dog())
make_sound(Cat())
```

```python
# Abstraction

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

r = Rectangle(5, 3)
print(r.area())
```

```python

class product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
        
    def __str__(self):
        return f"{self.name} ({self.price:,} rs)"
    
    def __repr__(self):
        return f"product({self.name!r}, {self.price})"
    
    def __eq__(self,other):
        same_name = self.name == other.name
        return same_name and self.price == other.price
     
    @property    # reads like attribute, computed every time
    def with_gst(self):
        return round(self.price * 1.18)
    

a = product("Earbuds",1299)
print(a)

cart = [product("vivo", 2002), product("case", 233)]
print(cart)

b = product("Earbuds",1299)
print(a==b)
print(a is b)

print(a.with_gst)
a.price = 999
print(a.with_gst)

```

```python
# Interview example: Valid parentheses

def is_valid(s):
    stack = []
    pairs = {")": "(", "]": "[", "}": "{"}

    for ch in s:
        if ch in "([{":
            stack.append(ch)
        elif ch in pairs:
            if not stack or stack.pop() != pairs[ch]:
                return False

    return len(stack) == 0

print(is_valid("({[]})"))  # True
print(is_valid("([)]"))    # False
```

```python
# Interview example: Two Sum (sets)
# Find two numbers whose sum equals a target.

def two_sum(nums, target):
    seen = {}

    for i, num in enumerate(nums):
        needed = target - num

        if needed in seen:
            return [seen[needed], i]

        seen[num] = i

    return []

print(two_sum([2, 7, 11, 15], 9))
# [0, 1]

#Time complexity: \(O(n)\) average. Space complexity: \(O(n)\).S
```

```python
# Breath  BFS (Graph Traversal)

from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])

    while queue:
        node = queue.popleft()
        print(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

bfs(graph, "A")
```

```python
# Depth  DFS (Graph Traversal)

def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start)

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

dfs(graph, "A")
```

```python
# Tries

class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root

        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]

        node.is_end = True

    def search(self, word):
        node = self.root

        for char in word:
            if char not in node.children:
                return False
            node = node.children[char]

        return node.is_end

trie = Trie()
trie.insert("cat")
trie.insert("car")

print(trie.search("cat"))  # True
print(trie.search("can"))  # False
```

### Searching - ( Binary , Linear ) 

```python
# Binary Search

def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1

numbers = [10, 20, 30, 40, 50, 60, 70]

result = binary_search(numbers, 60)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")
```

```python
# Linear search

def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

numbers = [10, 20, 30, 40, 50]

result = linear_search(numbers, 40)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")
```

### Sorting - ( Bubble, Selection, Insertion, Merge, Quick, Heap )

```python
# Bubble sort

def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr

numbers = [5, 3, 8, 1, 2]
print(bubble_sort(numbers))
```

```python
# Selection sort

def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr

numbers = [64, 25, 12, 22, 11]
print(selection_sort(numbers))
```

```python
# Insertion sort

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr

numbers = [5, 2, 4, 6, 1, 3]
print(insertion_sort(numbers))
```

```python
# Merge sort

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

numbers = [38, 27, 43, 3, 9, 82, 10]
print(merge_sort(numbers))
```

```python
# Quick sort

def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    smaller = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    greater = [x for x in arr if x > pivot]

    return quick_sort(smaller) + equal + quick_sort(greater)

numbers = [10, 7, 8, 9, 1, 5]
print(quick_sort(numbers))
```

```python
# Heap sort

def heap_sort(arr):
    import heapq

    # A min-heap; extract elements in ascending order.
    heap = arr.copy()
    heapq.heapify(heap)

    result = []
    while heap:
        result.append(heapq.heappop(heap))

    return result

numbers = [12, 11, 13, 5, 6, 7]
print(heap_sort(numbers))
```

# Two - Pointers

```python

def two_sum_sorted(arr, target):
    left = 0
    right = len(arr) - 1

    while left < right:
        total = arr[left] + arr[right]

        if total == target:
            return [left, right]

        elif total < target:
            left += 1

        else:
            right -= 1

    return []

numbers = [1, 2, 4, 6, 8, 9]
print(two_sum_sorted(numbers, 10))
```

```python
# Two Pointers (Check whether a string is a palindrome)

def is_palindrome(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome("madam"))
print(is_palindrome("hello"))
```

# Sliding Window

```python

def max_sum_subarray(arr, k):
    if k <= 0 or k > len(arr):
        raise ValueError("k must be between 1 and len(arr)")

    window_sum = sum(arr[:k])
    max_sum = window_sum

    for right in range(k, len(arr)):
        window_sum += arr[right]       # Add new element
        window_sum -= arr[right - k]   # Remove old element

        max_sum = max(max_sum, window_sum)

    return max_sum

numbers = [2, 1, 5, 1, 3, 2]
print(max_sum_subarray(numbers, 3))
```

```python
# Sliding Window (Longest substring without repeating characters)

def longest_unique_substring(s):
    left = 0
    last_seen = {}
    longest = 0

    for right, char in enumerate(s):
        if char in last_seen and last_seen[char] >= left:
            left = last_seen[char] + 1

        last_seen[char] = right
        longest = max(longest, right - left + 1)

    return longest

print(longest_unique_substring("abcabcbb"))
```

# Recursion

```python
# Recursion ( Factorial )

def factorial(n):
    if n == 0:          # Base case
        return 1

    return n * factorial(n - 1)

print(factorial(5))

# Time complexity / Auxiliary space: O(n)
```

```python
# Recursion ( Sum of numbers )

def sum_numbers(n):
    if n == 0:
        return 0

    return n + sum_numbers(n - 1)

print(sum_numbers(5))
```

# Dynamic Programming (DP)

```python

def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))

# Time: O(n).
# Auxiliary space: O(1).
```
